# VisionPulse — AI Visual Assistant for People with Low Vision
## Convolutional Neural Network (CNN) Training & Evaluation Notebook

This notebook demonstrates the complete Deep Learning pipeline for VisionPulse:
1. Dataset loading & visual data preprocessing
2. Data augmentation
3. CNN architecture construction (Conv2D, ReLU, MaxPooling, Flatten, Dense, Dropout, Softmax)
4. Model compilation & training with Adam optimizer
5. Loss & Accuracy performance curve plotting
6. Model evaluation (Accuracy, Precision, Recall, F1-Score, Confusion Matrix)
7. Sample image inference & spoken visual description generation

```python
import os
import sys
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from keras import layers, models
from sklearn.metrics import classification_report, confusion_matrix

# Ensure project root is imported
sys.path.insert(0, os.path.abspath('..'))
from src.config import INPUT_SHAPE, CLASS_NAMES, BATCH_SIZE, EPOCHS, LEARNING_RATE, MODEL_PATH
from src.model import build_cnn, compile_model
from src.description import generate_description
from src.speech import speak_description

print(f"TensorFlow Version: {tf.__version__}")
```

### Step 1: Load and Preprocess Dataset (CIFAR-10 / Custom Objects)

```python
# Load standard benchmark visual object dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Normalize pixel values [0, 255] -> [0.0, 1.0]
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# One-hot encode class target vectors
y_train_cat = tf.keras.utils.to_categorical(y_train, num_classes=len(CLASS_NAMES))
y_test_cat = tf.keras.utils.to_categorical(y_test, num_classes=len(CLASS_NAMES))

# 20% Validation Split
val_split = int(len(x_train) * 0.2)
x_val, y_val_cat = x_train[:val_split], y_train_cat[:val_split]
x_train_split, y_train_cat_split = x_train[val_split:], y_train_cat[val_split:]

print(f"Training set shape:   {x_train_split.shape}")
print(f"Validation set shape: {x_val.shape}")
print(f"Testing set shape:    {x_test.shape}")
print(f"Classes ({len(CLASS_NAMES)}): {CLASS_NAMES}")
```

### Step 2: Build & Compile CNN Model

```python
model = build_cnn(input_shape=INPUT_SHAPE, num_classes=len(CLASS_NAMES))
model = compile_model(model, learning_rate=LEARNING_RATE)
model.summary()
```

### Step 3: Train CNN Model with Callbacks

```python
callbacks_list = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    tf.keras.callbacks.ModelCheckpoint(MODEL_PATH, monitor='val_accuracy', save_best_only=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3)
]

history = model.fit(
    x_train_split, y_train_cat_split,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_data=(x_val, y_val_cat),
    callbacks=callbacks_list
)
```

### Step 4: Visualize Training Curves (Accuracy & Loss)

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy curve
ax1.plot(history.history['accuracy'], label='Train Accuracy', color='#00E5FF', linewidth=2)
ax1.plot(history.history['val_accuracy'], label='Val Accuracy', color='#FFD700', linewidth=2)
ax1.set_title('CNN Model Accuracy')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Accuracy')
ax1.legend()
ax1.grid(True)

# Loss curve
ax2.plot(history.history['loss'], label='Train Loss', color='#FF5252', linewidth=2)
ax2.plot(history.history['val_loss'], label='Val Loss', color='#AB47BC', linewidth=2)
ax2.set_title('CNN Model Loss')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Loss')
ax2.legend()
ax2.grid(True)

plt.suptitle('VisionPulse Training Performance', fontsize=16)
plt.show()
```

### Step 5: Evaluate Model Performance on Test Set

```python
y_preds_prob = model.predict(x_test)
y_preds = np.argmax(y_preds_prob, axis=1)
y_true = y_test.flatten()

print("Classification Report:")
print(classification_report(y_true, y_preds, target_names=CLASS_NAMES))
```

### Step 6: Sample Inference & Natural Language Spoken Description Generation

```python
sample_idx = 0
test_sample = np.expand_dims(x_test[sample_idx], axis=0)
sample_probs = model.predict(test_sample)[0]

sorted_indices = np.argsort(sample_probs)[::-1]
top_predictions = [
    {"class": CLASS_NAMES[i], "confidence": round(float(sample_probs[i]) * 100, 2)}
    for i in sorted_indices[:3]
]

desc = generate_description(top_predictions)
print("=== Generated Description ===")
print(desc['description'])
print("\nSpoken TTS Text:", desc['spoken_text'])
```
